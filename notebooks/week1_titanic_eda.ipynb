```python
## ==========================================
# DATA CLEANING + EDA
# ==========================================

# Check original data
print("Original shape:", df.shape)

# ------------------------------------------
# 1. MISSING VALUES
# ------------------------------------------

print("\nMissing values before cleaning:")
print(df.isnull().sum())

# ------------------------------------------
# 2. MISSING VALUE VISUALIZATION
# ------------------------------------------

plt.figure(figsize=(10, 5))

sns.heatmap(df.isnull(), cbar=False)

plt.title("Missing Values in Titanic Dataset")
plt.tight_layout()

plt.savefig("../visualizations/01_missing_values.png")
plt.show()


# ------------------------------------------
# 3. DATA CLEANING
# ------------------------------------------

# Standardize categorical columns
for col in ["sex", "embarked", "class", "who", "alive", "embark_town"]:
    df[col] = df[col].astype("string").str.strip()


# Convert numerical columns
numeric_cols = [
    "survived",
    "pclass",
    "age",
    "sibsp",
    "parch",
    "fare"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")


# ------------------------------------------
# 4. HANDLE AGE
# ------------------------------------------

print("\nMissing age before:", df["age"].isnull().sum())

df["age"] = df.groupby(
    ["pclass", "sex"]
)["age"].transform(
    lambda x: x.fillna(x.median())
)

print("Missing age after:", df["age"].isnull().sum())


# ------------------------------------------
# 5. HANDLE EMBARKED
# ------------------------------------------

print("\nMissing embarked before:",
      df["embarked"].isnull().sum())

df["embarked"] = df["embarked"].fillna(
    df["embarked"].mode()[0]
)

print("Missing embarked after:",
      df["embarked"].isnull().sum())


# ------------------------------------------
# 6. HANDLE DECK
# ------------------------------------------

# deck has very high missing values.
# Instead of guessing deck values, remove it.

print("\nMissing deck:")
print(df["deck"].isnull().sum())

df = df.drop(columns=["deck"])


# ------------------------------------------
# 7. REMOVE DUPLICATES
# ------------------------------------------

print("\nDuplicates before:",
      df.duplicated().sum())

df = df.drop_duplicates().reset_index(drop=True)

print("Duplicates after:",
      df.duplicated().sum())


# ------------------------------------------
# 8. FINAL VALIDATION
# ------------------------------------------

print("\n================================")
print("CLEANED DATASET")
print("================================")

print("Shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nData types:")
print(df.dtypes)


# Save cleaned dataset
df.to_csv(
    "../data/titanic_cleaned.csv",
    index=False
)


# ==========================================
# EDA
# ==========================================

# ------------------------------------------
# 9. SURVIVAL BY SEX
# ------------------------------------------

sex_survival = (
    df.groupby("sex")["survived"].mean() * 100
)

plt.figure(figsize=(7, 5))

ax = sns.barplot(
    x=sex_survival.index,
    y=sex_survival.values
)

plt.title("Survival Rate by Sex")
plt.xlabel("Sex")
plt.ylabel("Survival Rate (%)")
plt.ylim(0, 100)

for i, value in enumerate(sex_survival.values):
    ax.text(
        i,
        value + 2,
        f"{value:.1f}%",
        ha="center"
    )

plt.tight_layout()

plt.savefig(
    "../visualizations/02_survival_by_sex.png"
)

plt.show()


# ------------------------------------------
# 10. SURVIVAL BY CLASS
# ------------------------------------------

class_survival = (
    df.groupby("pclass")["survived"].mean() * 100
)

plt.figure(figsize=(7, 5))

ax = sns.barplot(
    x=class_survival.index.astype(str),
    y=class_survival.values
)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate (%)")
plt.ylim(0, 100)

for i, value in enumerate(class_survival.values):
    ax.text(
        i,
        value + 2,
        f"{value:.1f}%",
        ha="center"
    )

plt.tight_layout()

plt.savefig(
    "../visualizations/03_survival_by_class.png"
)

plt.show()


# ------------------------------------------
# 11. AGE DISTRIBUTION
# ------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    df["age"],
    bins=20,
    kde=True
)

plt.title("Age Distribution After Imputation")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.tight_layout()

plt.savefig(
    "../visualizations/04_age_distribution.png"
)

plt.show()


# ------------------------------------------
# 12. CORRELATION MATRIX
# ------------------------------------------

corr_columns = [
    "survived",
    "pclass",
    "age",
    "sibsp",
    "parch",
    "fare"
]

plt.figure(figsize=(8, 6))

sns.heatmap(
    df[corr_columns].corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Matrix")

plt.tight_layout()

plt.savefig(
    "../visualizations/05_correlation_matrix.png"
)

plt.show()


# ==========================================
# FINAL INSIGHTS
# ==========================================

overall_survival = (
    df["survived"].mean() * 100
)

print("\n================================")
print("FINAL EDA INSIGHTS")
print("================================")

print(
    f"\nOverall survival rate: "
    f"{overall_survival:.2f}%"
)

print("\nSurvival by sex:")
print(sex_survival.round(2))

print("\nSurvival by passenger class:")
print(class_survival.round(2))

print(
    f"\nMedian age: "
    f"{df['age'].median():.2f}"
)

print(
    f"Median fare: "
    f"{df['fare'].median():.2f}"
)

print("\nFinal shape:", df.shape)

print("\nRemaining missing values:")
print(df.isnull().sum())

print("\n✅ PROJECT ANALYSIS COMPLETED!")
```
